```python
import pyiqa
import numpy as np
import os

niqe_metric = pyiqa.create_metric('niqe')
brisque_metric = pyiqa.create_metric('brisque')

folder = "gan_enhanced_results"

niqe_scores = []
brisque_scores = []

for img_name in os.listdir(folder):

    if not img_name.lower().endswith(('.png','.jpg','.jpeg','.bmp')):
        continue

    path = os.path.join(folder, img_name)

    score_niqe = niqe_metric(path)
    score_brisque = brisque_metric(path)

    niqe_scores.append(score_niqe.item())
    brisque_scores.append(score_brisque.item())

print("Average NIQE:", np.mean(niqe_scores))
print("Average BRISQUE:", np.mean(brisque_scores))
```

```python
import pyiqa
import numpy as np
import os
import cv2

niqe_metric = pyiqa.create_metric('niqe')
brisque_metric = pyiqa.create_metric('brisque')

folder = "preprocessed/ExDark"

niqe_scores = []
brisque_scores = []

for img_name in os.listdir(folder):

    if not img_name.lower().endswith(('.png','.jpg','.jpeg','.bmp')):
        continue

    path = os.path.join(folder, img_name)

    img = cv2.imread(path)

    # Skip corrupted images
    if img is None:
        print("Skipping corrupted:", img_name)
        continue

    # Skip very small images
    h, w = img.shape[:2]
    if h < 64 or w < 64:
        print("Skipping small image:", img_name)
        continue

    try:
        niqe_scores.append(niqe_metric(path).item())
        brisque_scores.append(brisque_metric(path).item())

    except Exception as e:
        print("Skipping problematic image:", img_name)

print("\nAverage NIQE:", np.mean(niqe_scores))
print("Average BRISQUE:", np.mean(brisque_scores))
```

```python
from ultralytics import YOLO
import cv2
import os

model = YOLO("yolov8n.pt")  

# Paths
low_light_dir = "preprocessed/ExDark"
enhanced_dir = "gan_enhanced_results"

output_low = "detections_low"
output_enhanced = "detections_enhanced"

os.makedirs(output_low, exist_ok=True)
os.makedirs(output_enhanced, exist_ok=True)


print("Running detection on enhanced images...")

for img_name in os.listdir(enhanced_dir):

    img_path = os.path.join(enhanced_dir, img_name)

    # Run detection
    results = model(img_path)

    # Save image with bounding boxes
    result_img = results[0].plot()
    
    for box in results[0].boxes:
        conf = float(box.conf)
        cls = int(box.cls)
        label = model.names[cls]

        print(f"{img_name} -> {label} : {conf:.2f}")

    save_path = os.path.join(output_enhanced, img_name)
    cv2.imwrite(save_path, result_img)

print("Enhanced image detection completed.")
```
